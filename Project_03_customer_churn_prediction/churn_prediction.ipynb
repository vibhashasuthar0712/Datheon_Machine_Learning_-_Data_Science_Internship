```python

# ============================================================
# PROJECT: TELECOM CUSTOMER CHURN PREDICTION
# DATASET: IBM TELCO CUSTOMER CHURN
# ============================================================


# ============================================================
# 1. IMPORT REQUIRED LIBRARIES
# ============================================================

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    roc_curve
)

import warnings
warnings.filterwarnings('ignore')

print("All libraries imported successfully!")


# ============================================================
# 2. LOAD DATASET
# ============================================================

df = pd.read_csv("customer_churn.csv")

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

display(df.head())


# ============================================================
# 3. INITIAL DATA UNDERSTANDING
# ============================================================

# Check dataset information
df.info()

# Statistical summary
display(df.describe(include='all'))

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Check duplicate rows
print("\nDuplicate Rows:", df.duplicated().sum())


# ============================================================
# 4. DATA CLEANING AND PREPROCESSING
# ============================================================

# Remove customerID because it is only a unique identifier
df.drop(columns=['customerID'], inplace=True, errors='ignore')

# Convert TotalCharges into numeric format
# Blank or invalid values will be converted into NaN
df['TotalCharges'] = pd.to_numeric(
    df['TotalCharges'].astype(str).str.strip(),
    errors='coerce'
)

# Handle missing TotalCharges values
# Customers with tenure = 0 generally have blank TotalCharges
df['TotalCharges'] = df['TotalCharges'].fillna(0)

# Verify data types after cleaning
print("\nData Types After Cleaning:")
print(df.dtypes)

# Check missing values after cleaning
print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

print("\nCleaned Dataset Shape:", df.shape)


# ============================================================
# 5. EXPLORATORY DATA ANALYSIS (EDA)
# ============================================================

# Check target variable distribution
print("\nChurn Distribution:")
print(df['Churn'].value_counts())

# Check churn percentage
print("\nChurn Percentage:")
print(
    (df['Churn'].value_counts(normalize=True) * 100).round(2)
)

# Visualize target distribution
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x='Churn')
plt.title("Customer Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Number of Customers")
plt.show()


# ============================================================
# 6. CHURN ANALYSIS BY IMPORTANT CATEGORICAL FEATURES
# ============================================================

# Analyze churn percentage for different customer groups

categorical_analysis = [
    'Contract',
    'InternetService',
    'PaymentMethod',
    'OnlineSecurity',
    'TechSupport',
    'PaperlessBilling',
    'SeniorCitizen',
    'gender',
    'Partner'
]

for col in categorical_analysis:

    print("\n" + "=" * 50)
    print(f"{col} vs Churn Percentage")
    print("=" * 50)

    churn_percentage = pd.crosstab(
        df[col],
        df['Churn'],
        normalize='index'
    ) * 100

    display(churn_percentage.round(2))


# ============================================================
# 7. NUMERICAL FEATURE ANALYSIS
# ============================================================

# Compare numerical features based on churn status

print("\nAverage Customer Tenure:")
print(df.groupby('Churn')['tenure'].mean().round(2))

print("\nAverage Monthly Charges:")
print(df.groupby('Churn')['MonthlyCharges'].mean().round(2))

print("\nAverage Total Charges:")
print(df.groupby('Churn')['TotalCharges'].mean().round(2))


# ============================================================
# 8. CORRELATION ANALYSIS
# ============================================================

# Create a separate copy for correlation analysis
# Churn_Encoded will NOT be used in model training

df_corr = df.copy()

# Convert Churn into numerical values only for correlation
df_corr['Churn_Encoded'] = df_corr['Churn'].map({
    'No': 0,
    'Yes': 1
})

# Select numerical columns
numerical_corr = df_corr.select_dtypes(include=np.number)

# Calculate correlation with Churn
print("\nCorrelation with Churn:")
print(
    numerical_corr.corr()['Churn_Encoded']
    .sort_values(ascending=False)
)

# Correlation heatmap
plt.figure(figsize=(10, 7))

sns.heatmap(
    numerical_corr.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Correlation Heatmap")
plt.show()


# ============================================================
# 9. DEFINE FEATURES AND TARGET
# ============================================================

# Remove target column from input features
X = df.drop(columns=['Churn'])

# Encode target variable
y = df['Churn'].map({
    'No': 0,
    'Yes': 1
})

# Verify that the target has been encoded correctly
print("\nTarget Distribution:")
print(y.value_counts())

# Verify feature and target shapes
print("\nX Shape:", X.shape)
print("Y Shape:", y.shape)

# Check feature columns
print("\nInput Features:")
print(X.columns.tolist())


# ============================================================
# 10. IDENTIFY NUMERICAL AND CATEGORICAL FEATURES
# ============================================================

# SeniorCitizen is a categorical indicator (0 or 1),
# so include it in categorical preprocessing.

numerical_features = [
    'tenure',
    'MonthlyCharges',
    'TotalCharges'
]

categorical_features = [
    col for col in X.columns
    if col not in numerical_features
]

print("\nNumerical Features:", numerical_features)
print("Number of Numerical Features:", len(numerical_features))

print("\nCategorical Features:", categorical_features)
print("Number of Categorical Features:", len(categorical_features))


# ============================================================
# 11. TRAIN-TEST SPLIT
# ============================================================

# Split data into 80% training and 20% testing
# Stratify maintains the original class distribution

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining Shape:", X_train.shape)
print("Testing Shape:", X_test.shape)

print("\nTraining Target Distribution:")
print(y_train.value_counts(normalize=True).round(3))

print("\nTesting Target Distribution:")
print(y_test.value_counts(normalize=True).round(3))


# ============================================================
# 12. HANDLE CLASS IMBALANCE
# ============================================================

# Calculate class weight ratio using training data only
# This helps the models give more importance to the minority class

class_counts = y_train.value_counts()

scale_pos_weight = (
    class_counts[0] / class_counts[1]
)

print("\nClass Distribution in Training Data:")
print(class_counts)

print("\nXGBoost scale_pos_weight:", scale_pos_weight)


# ============================================================
# 13. CREATE PREPROCESSING PIPELINE
# ============================================================

# Numerical preprocessing:
# Fill missing values using median

numeric_transformer = Pipeline(
    steps=[
        ('imputer', SimpleImputer(strategy='median'))
    ]
)

# Categorical preprocessing:
# Fill missing values using most frequent value
# Convert categorical features into numerical format using OneHotEncoder

categorical_transformer = Pipeline(
    steps=[
        ('imputer', SimpleImputer(strategy='most_frequent')),
        ('onehot', OneHotEncoder(handle_unknown='ignore'))
    ]
)

# Combine numerical and categorical preprocessing

preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numerical_features),
        ('cat', categorical_transformer, categorical_features)
    ]
)

print("\nPreprocessing pipeline created successfully!")


# ============================================================
# 14. INITIALIZE MACHINE LEARNING MODELS
# ============================================================

# Random Forest Classifier

rf_model = RandomForestClassifier(
    n_estimators=300,
    max_depth=None,
    min_samples_split=2,
    random_state=42,
    class_weight='balanced',
    n_jobs=-1
)

# XGBoost Classifier

xgb_model = XGBClassifier(
    n_estimators=300,
    max_depth=5,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=scale_pos_weight,
    eval_metric='logloss',
    random_state=42,
    n_jobs=-1
)

print("\nRandom Forest and XGBoost initialized!")


# ============================================================
# 15. CREATE COMPLETE MODEL PIPELINES
# ============================================================

# Preprocessing and model training are combined
# so preprocessing is fitted separately inside each CV fold.

rf_pipeline = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('classifier', rf_model)
    ]
)

xgb_pipeline = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('classifier', xgb_model)
    ]
)

print("\nBoth model pipelines created successfully!")


# ============================================================
# 16. TRAIN BOTH MODELS
# ============================================================

# Train Random Forest
rf_pipeline.fit(X_train, y_train)

print("\nRandom Forest training completed!")

# Train XGBoost
xgb_pipeline.fit(X_train, y_train)

print("XGBoost training completed!")


# ============================================================
# 17. STRATIFIED 5-FOLD CROSS-VALIDATION
# ============================================================

# Use StratifiedKFold to maintain class distribution
# in each validation fold

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# Random Forest cross-validation
rf_cv_scores = cross_val_score(
    rf_pipeline,
    X_train,
    y_train,
    cv=cv,
    scoring='roc_auc',
    n_jobs=1
)

# XGBoost cross-validation
xgb_cv_scores = cross_val_score(
    xgb_pipeline,
    X_train,
    y_train,
    cv=cv,
    scoring='roc_auc',
    n_jobs=1
)

print("\nRandom Forest CV Scores:")
print(rf_cv_scores)

print("Random Forest Mean CV ROC-AUC:")
print(round(rf_cv_scores.mean(), 4))


print("\nXGBoost CV Scores:")
print(xgb_cv_scores)

print("XGBoost Mean CV ROC-AUC:")
print(round(xgb_cv_scores.mean(), 4))


# ============================================================
# 18. MODEL PREDICTIONS
# ============================================================

# Random Forest predictions
rf_predictions = rf_pipeline.predict(X_test)

rf_probabilities = rf_pipeline.predict_proba(X_test)[:, 1]


# XGBoost predictions
xgb_predictions = xgb_pipeline.predict(X_test)

xgb_probabilities = xgb_pipeline.predict_proba(X_test)[:, 1]


# ============================================================
# 19. MODEL EVALUATION
# ============================================================

# Create a function to evaluate model performance

def evaluate_model(model_name, y_true, predictions, probabilities):

    print("\n" + "=" * 55)
    print(model_name)
    print("=" * 55)

    accuracy = accuracy_score(y_true, predictions)

    precision = precision_score(
        y_true,
        predictions,
        zero_division=0
    )

    recall = recall_score(
        y_true,
        predictions,
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        predictions,
        zero_division=0
    )

    roc_auc = roc_auc_score(
        y_true,
        probabilities
    )

    print("Accuracy:", round(accuracy, 4))
    print("Precision:", round(precision, 4))
    print("Recall:", round(recall, 4))
    print("F1 Score:", round(f1, 4))
    print("ROC-AUC:", round(roc_auc, 4))

    print("\nClassification Report:")

    print(
        classification_report(
            y_true,
            predictions,
            target_names=['No Churn', 'Churn'],
            zero_division=0
        )
    )

    return {
        'Model': model_name,
        'Accuracy': accuracy,
        'Precision': precision,
        'Recall': recall,
        'F1 Score': f1,
        'ROC-AUC': roc_auc
    }


# Evaluate Random Forest
rf_results = evaluate_model(
    "Random Forest",
    y_test,
    rf_predictions,
    rf_probabilities
)

# Evaluate XGBoost
xgb_results = evaluate_model(
    "XGBoost",
    y_test,
    xgb_predictions,
    xgb_probabilities
)


# ============================================================
# 20. CONFUSION MATRICES
# ============================================================

# Random Forest confusion matrix

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

ConfusionMatrixDisplay.from_predictions(
    y_test,
    rf_predictions,
    display_labels=['No Churn', 'Churn'],
    cmap='Blues',
    ax=axes[0]
)

axes[0].set_title("Random Forest Confusion Matrix")


# XGBoost confusion matrix

ConfusionMatrixDisplay.from_predictions(
    y_test,
    xgb_predictions,
    display_labels=['No Churn', 'Churn'],
    cmap='Greens',
    ax=axes[1]
)

axes[1].set_title("XGBoost Confusion Matrix")

plt.tight_layout()
plt.show()


# ============================================================
# 21. ROC-AUC CURVE COMPARISON
# ============================================================

# Calculate ROC curve for Random Forest

rf_fpr, rf_tpr, _ = roc_curve(
    y_test,
    rf_probabilities
)

# Calculate ROC curve for XGBoost

xgb_fpr, xgb_tpr, _ = roc_curve(
    y_test,
    xgb_probabilities
)

# Plot both ROC curves

plt.figure(figsize=(9, 6))

plt.plot(
    rf_fpr,
    rf_tpr,
    label=f"Random Forest (AUC = {rf_results['ROC-AUC']:.4f})"
)

plt.plot(
    xgb_fpr,
    xgb_tpr,
    label=f"XGBoost (AUC = {xgb_results['ROC-AUC']:.4f})"
)

# Random classifier reference line
plt.plot(
    [0, 1],
    [0, 1],
    linestyle='--',
    label='Random Classifier'
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC-AUC Curve Comparison")

plt.legend()
plt.grid()
plt.show()


# ============================================================
# 22. FEATURE IMPORTANCE ANALYSIS
# ============================================================

# Get transformed feature names from the fitted preprocessor

feature_names = (
    rf_pipeline
    .named_steps['preprocessor']
    .get_feature_names_out()
)

# Random Forest feature importance

rf_importances = (
    rf_pipeline
    .named_steps['classifier']
    .feature_importances_
)

rf_feature_importance = pd.Series(
    rf_importances,
    index=feature_names
).sort_values(ascending=False)

print("\nTop 15 Random Forest Features:")
print(rf_feature_importance.head(15))


# XGBoost feature importance

xgb_importances = (
    xgb_pipeline
    .named_steps['classifier']
    .feature_importances_
)

xgb_feature_importance = pd.Series(
    xgb_importances,
    index=feature_names
).sort_values(ascending=False)

print("\nTop 15 XGBoost Features:")
print(xgb_feature_importance.head(15))


# ============================================================
# 23. VISUALIZE TOP 10 IMPORTANT FEATURES
# ============================================================

# Random Forest top 10 features

plt.figure(figsize=(10, 6))

rf_feature_importance.head(10).sort_values().plot(
    kind='barh'
)

plt.title("Top 10 Random Forest Features")
plt.xlabel("Feature Importance")
plt.ylabel("Features")
plt.tight_layout()
plt.show()


# XGBoost top 10 features

plt.figure(figsize=(10, 6))

xgb_feature_importance.head(10).sort_values().plot(
    kind='barh'
)

plt.title("Top 10 XGBoost Features")
plt.xlabel("Feature Importance")
plt.ylabel("Features")
plt.tight_layout()
plt.show()


# ============================================================
# 24. FINAL MODEL COMPARISON
# ============================================================

# Combine evaluation results into a DataFrame

results_df = pd.DataFrame([
    rf_results,
    xgb_results
])

# Add cross-validation ROC-AUC scores

results_df['CV ROC-AUC'] = [
    rf_cv_scores.mean(),
    xgb_cv_scores.mean()
]

print("\nFINAL MODEL COMPARISON")
print("=" * 65)

display(
    results_df.round(4)
)


# Select the best model using test ROC-AUC

best_model_name = results_df.loc[
    results_df['ROC-AUC'].idxmax(),
    'Model'
]

best_roc_auc = results_df['ROC-AUC'].max()

print("\nBEST MODEL:", best_model_name)
print("Test ROC-AUC:", round(best_roc_auc, 4))


# ============================================================
# 25. TOP 10 IMPORTANT FEATURES OF BEST MODEL
# ============================================================

if best_model_name == 'XGBoost':

    best_feature_importance = xgb_feature_importance

else:

    best_feature_importance = rf_feature_importance


print("\nTOP 10 IMPORTANT FEATURES")
print("=" * 55)

top_10_features = best_feature_importance.head(10)

print(top_10_features)


# ============================================================
# 26. BUSINESS INSIGHTS
# ============================================================

# Overall customer churn rate

churn_rate = (
    df['Churn'].eq('Yes').mean() * 100
)

print("\nOverall Customer Churn Rate:")
print(round(churn_rate, 2), "%")


# Churn percentage by contract type

print("\nChurn Percentage by Contract:")

contract_churn = (
    pd.crosstab(
        df['Contract'],
        df['Churn'],
        normalize='index'
    )['Yes'] * 100
)

print(contract_churn.round(2))


# Average customer tenure

print("\nAverage Customer Tenure:")

print(
    df.groupby('Churn')['tenure']
    .mean()
    .round(2)
)


# Average monthly charges

print("\nAverage Monthly Charges:")

print(
    df.groupby('Churn')['MonthlyCharges']
    .mean()
    .round(2)
)


# Churn percentage by payment method

print("\nChurn Percentage by Payment Method:")

payment_churn = (
    pd.crosstab(
        df['PaymentMethod'],
        df['Churn'],
        normalize='index'
    )['Yes'] * 100
)

print(payment_churn.round(2))


# Churn percentage by online security

print("\nChurn Percentage by Online Security:")

security_churn = (
    pd.crosstab(
        df['OnlineSecurity'],
        df['Churn'],
        normalize='index'
    )['Yes'] * 100
)

print(security_churn.round(2))


# ============================================================
# 27. FINAL BUSINESS INTERPRETATION
# ============================================================

print("\nFINAL BUSINESS INTERPRETATION")
print("=" * 65)

print("\n1. CUSTOMER CHURN:")
print(
    f"Overall customer churn rate is {churn_rate:.2f}%."
)

print("\n2. CONTRACT TYPE:")
print(
    f"Month-to-month contract churn rate is "
    f"{contract_churn.get('Month-to-month', 0):.2f}%."
)
print(
    "The company can focus on improving retention "
    "among month-to-month customers."
)

print("\n3. CUSTOMER TENURE:")

tenure_means = df.groupby('Churn')['tenure'].mean()

print(
    f"Average tenure of retained customers: "
    f"{tenure_means.get('No', 0):.2f} months."
)

print(
    f"Average tenure of churned customers: "
    f"{tenure_means.get('Yes', 0):.2f} months."
)

print(
    "Early customer engagement may help improve retention."
)

print("\n4. MONTHLY CHARGES:")

monthly_means = df.groupby('Churn')['MonthlyCharges'].mean()

print(
    f"Average monthly charges for churned customers: "
    f"{monthly_means.get('Yes', 0):.2f}."
)

print(
    f"Average monthly charges for retained customers: "
    f"{monthly_means.get('No', 0):.2f}."
)

print(
    "Reviewing pricing and service value may help "
    "address customer dissatisfaction."
)

print("\n5. PAYMENT METHOD:")

print(
    f"Electronic check churn rate is "
    f"{payment_churn.get('Electronic check', 0):.2f}%."
)

print(
    "The company can investigate the customer experience "
    "associated with this payment method."
)

print("\n6. MODEL PERFORMANCE:")

print(f"Selected model: {best_model_name}")

print(f"Test ROC-AUC score: {best_roc_auc:.4f}")

print("Required ROC-AUC target: 0.80")

if best_roc_auc >= 0.80:

    print("RESULT: ROC-AUC target achieved!")

else:

    print("RESULT: ROC-AUC target not achieved.")

print("\nProject analysis completed successfully!")


# ============================================================
# 28. FINAL PROJECT SUMMARY
# ============================================================

print("\nFINAL PROJECT SUMMARY")
print("=" * 60)

print("Project: Telecom Customer Churn Prediction")
print("Dataset: IBM Telco Customer Churn")
print("Total Records:", len(df))
print("Input Features:", X.shape[1])
print("Models Used: Random Forest and XGBoost")
print("Class Imbalance Handling: Class Weights")
print("Cross Validation: 5-Fold Stratified Cross Validation")
print("Best Model:", best_model_name)
print("Best Test ROC-AUC:", round(best_roc_auc, 4))

print("\nModel Comparison:")

display(results_df.round(4))

print("\nFinal Project Execution Completed!")
```
